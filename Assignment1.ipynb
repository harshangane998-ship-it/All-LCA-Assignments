```python
from datetime import date

x1 = int(input("enter year:"))
y1 = int(input("enter month:"))
z1 = int(input("enter day:"))
x2 = int(input("enter other person year:"))
y2 = int(input("enter other person month:"))
z2 = int(input("enter other person day:"))

t1 = date(x1 , y1 , z1)
t2 = date(x2 , y2 , z2)

print(abs((t2-t1).days))
```
