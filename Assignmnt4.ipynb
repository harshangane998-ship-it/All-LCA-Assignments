```python
matrix1 = [[1,4,5],
           [4,5,6]]

matrix2 = [[2,4,9],
           [8,4,6]]

result = [[0,0,0],
          [0,0,0]]

for i in range(len(matrix1)):
    for j in range(len(matrix1[0])):
        result[i][j] = matrix1[i][j] + matrix2[i][j]
print("matrix:1")
for row in matrix1:
    print(row)

print("\nmatrix2:")
for row in matrix2:
    print(row)

print("\nsum of matrix1 and matrix2:")
for row in result:
    print(row)
    
```
